```python
import matplotlib.pyplot as plt
import numpy as np
import json
```

```python
with open('./rosenbrock_data.json', 'r', encoding='utf-8') as file:
    rosenbrock_data = json.load(file)

with open('./rastrigin_data.json', 'r', encoding='utf-8') as file:
    rastrigin_data = json.load(file)
```

---
# Rosenbrock Function

```python
rb_best = np.array([x['best'] for x in rosenbrock_data])
rb_avg = np.array([x['avg'] for x in rosenbrock_data])

rb_best_mean = []
rb_best_errors = [[], []]
rb_avg_mean = []
rb_avg_errors = [[], []]
for i in range(len(rb_best[0])):
    collector_best = []
    collector_avg = []

    for j in range(len(rb_best)):
        collector_best.append(rb_best[j][i])
        collector_avg.append(rb_avg[j][i])

    rb_best_mean.append(sum(collector_best) / len(collector_best))
    # rb_best_errors.append(np.std(collector_best, ddof=1) / np.sqrt(np.size(collector_best)))
    rb_best_errors[0].append(abs(sum(collector_best) / len(collector_best) - min(collector_best)))
    rb_best_errors[1].append(abs(sum(collector_best) / len(collector_best) - max(collector_best)))
    rb_avg_mean.append(sum(collector_avg) / len(collector_avg))
    rb_avg_errors[0].append(abs(sum(collector_avg) / len(collector_avg) - min(collector_avg)))
    rb_avg_errors[1].append(abs(sum(collector_avg) / len(collector_avg) - max(collector_avg)))

plt.figure(figsize=(12, 8))
plt.errorbar(
    range(len(rb_best_mean)), 
    rb_best_mean, 
    yerr=rb_best_errors,
    fmt='none',
    capsize=5,
    capthick=2,
    elinewidth=1,
    c='black')
plt.plot(rb_best_mean, lw=2)

ax = plt.gca()
ax.set_ylabel("Fitness")
ax.set_xlabel("Iteration")
ax.set_title("Rosenbrock Function - Top Fitness")
```

```python
plt.figure(figsize=(12, 8))
plt.errorbar(
    range(len(rb_avg_mean)), 
    rb_avg_mean, 
    yerr=rb_avg_errors,
    fmt='none',
    capsize=5,
    capthick=2,
    elinewidth=1,
    c='black')
plt.plot(rb_avg_mean, c='tab:orange', lw=2)

ax = plt.gca()
ax.set_ylabel("Fitness")
ax.set_xlabel("Iteration")
ax.set_title("Rosenbrock Function - Average Fitness")
```
